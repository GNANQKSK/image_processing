```python
import numpy as np
import matplotlib.pyplot as plt

# Load image as NumPy array
image = plt.imread("/content/DORAMON.webp")
print(image.shape)

# -----------------------------
# Horizontal Flip
# -----------------------------
h_flip = np.fliplr(image)
plt.imshow(h_flip)
plt.title("Horizontal Flip")
plt.axis("off")
plt.show()

# -----------------------------
# Vertical Flip
# -----------------------------
v_flip = np.flipud(image)
plt.imshow(v_flip)
plt.title("Vertical Flip")
plt.axis("off")
plt.show()

# -----------------------------
# Rotate 90 degrees
# -----------------------------
rotate_90 = np.rot90(image)
plt.imshow(rotate_90)
plt.title("Rotate 90")
plt.axis("off")
plt.show()

# -----------------------------
# Brightness Increased
# -----------------------------
bright = np.clip(image + 50, 0, 255).astype(np.uint8)
plt.imshow(bright)
plt.title("Brightness Increased")
plt.axis("off")
plt.show()

# -----------------------------
# Gaussian Noise
# -----------------------------
noise = np.random.normal(0, 25, image.shape)
noisy = np.clip(image + noise, 0, 255).astype(np.uint8)
plt.imshow(noisy)
plt.title("Gaussian Noise")
plt.axis("off")
plt.show()

# -----------------------------
# Crop Image (Center crop)
# -----------------------------
h, w, _ = image.shape
cropped = image[h//4:3*h//4, w//4:3*w//4]
plt.imshow(cropped)
plt.title("Cropped Image")
plt.axis("off")
plt.show()

# -----------------------------
# Grayscale Image
# -----------------------------
if image.ndim == 3:
    gray = np.mean(image, axis=2).astype(np.uint8)
else:
    gray = image

plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()

# -----------------------------
# Binary Thresholding
# -----------------------------
threshold = 128
binary = np.where(gray > threshold, 255, 0).astype(np.uint8)
plt.imshow(binary, cmap="gray")
plt.title("Binary Thresholding")
plt.axis("off")
plt.show()
```
